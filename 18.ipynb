```python
import numpy as np
import pandas as pd
df = pd.read_csv(r"D:\Downloads\iris - iris.csv")
# match the manual's column / class names
df.columns=['sepal.length','sepal.width','petal.length','petal.width','variety']
df['variety']=df['variety'].str.capitalize()
```

```python
features = df.iloc[:,:-1].values
label = df.iloc[:,4].to_numpy(dtype=object)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
xtrain,xtest,ytrain,ytest=train_test_split(features,label,test_size=.2,random_state=1)
model_KNN=KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(xtrain,ytrain)
```

```python
print(model_KNN.score(xtrain,ytrain))
print(model_KNN.score(xtest,ytest))
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(label,model_KNN.predict(features))
```

```python
from sklearn.metrics import classification_report
print(classification_report(label,model_KNN.predict(features)))
```
